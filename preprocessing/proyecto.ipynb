### Pais de nacimiento

```python
# Paso 1: Normalizar
df['PAIS_NACIMIENTO_NORMALIZADO'] = (
    df['PAIS_NACIMIENTO']
    .str.lower()
    .str.strip()
)

# Corrección manual para errores raros
df['PAIS_NACIMIENTO_NORMALIZADO'] = df['PAIS_NACIMIENTO_NORMALIZADO'].replace({
    'guineasin datobissau': 'guinea bissau'
})

# Paso 2: Mapping dict
mapping_dict_pais = {
    'colombia': 'colombia',
    'venezuela': 'venezuela',
    'sin dato': 'sin dato',
    'no aplica': 'sin dato'
    # todo lo demás será "otros países"
}

# Paso 3: Mapear
df['PAIS_NACIMIENTO_CATEGORIA'] = df['PAIS_NACIMIENTO_NORMALIZADO'].map(mapping_dict_pais)

# Paso 4: Los que no están mapeados van como "otros países"
df['PAIS_NACIMIENTO_CATEGORIA'] = df['PAIS_NACIMIENTO_CATEGORIA'].fillna('otros países')

# Verificar resultado
print(df['PAIS_NACIMIENTO_CATEGORIA'].value_counts())
```

### Edad

```python
# 1. Convertir EDAD a numérico (manejar errores)
df['EDAD'] = pd.to_numeric(df['EDAD'], errors='coerce')

# 2. Definir función de categorización
def categorizar_edad(edad):
    if pd.isnull(edad):
        return 'sin dato'
    elif edad <= 11:
        return 'infancia'
    elif edad <= 17:
        return 'adolescencia'
    elif edad <= 25:
        return 'juventud temprana'
    elif edad <= 35:
        return 'juventud adulta'
    elif edad <= 50:
        return 'adultez'
    elif edad <= 65:
        return 'adulto mayor joven'
    else:
        return 'adulto mayor'

# 3. Aplicar función
df['EDAD_CATEGORIA'] = df['EDAD'].apply(categorizar_edad)

# 4. Verificar
print(df['EDAD_CATEGORIA'].value_counts())
```

```python
df['EDAD_CATEGORIA'].value_counts().sort_index().plot(kind='bar', figsize=(12,6))
plt.title('Delitos por hora del día')
plt.xlabel('Hora del día')
plt.ylabel('Número de delitos')
plt.show()
```
